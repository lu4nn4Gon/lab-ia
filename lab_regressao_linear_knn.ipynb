# Regressão Linear — Água dos Alpes


```python
import pandas as pd
```

```python
dados = pd.read_csv("/agua_alpes.csv")
dados
```

```python
X = dados[["BPt"]]
y = dados["Pressure"]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
modelo = LinearRegression()
```

```python
modelo.fit(X, y)
```

```python
modelo.coef_
```

```python
modelo.intercept_
```

```python
temperatura_200 = pd.DataFrame({"BPt": [200]})
modelo.predict(temperatura_200)
```

```python
temperatura_215 = pd.DataFrame({"BPt": [215]})
modelo.predict(temperatura_215)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
y_pred = modelo.predict(X)
```

```python
mae = mean_absolute_error(y, y_pred)

mae
```

```python
mse = mean_squared_error(y, y_pred)
mse
```

```python
r2 = r2_score(y, y_pred)
r2
```

```python
rmse = mse ** 0.5
rmse
```

```python
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
pressao_200 = modelo.predict(pd.DataFrame({"BPt": [200]}))[0]
pressao_215 = modelo.predict(pd.DataFrame({"BPt": [215]}))[0]

print("Pressão prevista para 200°F:", pressao_200)
print("Pressão prevista para 215°F:", pressao_215)
```

# Regressão Linear — Livros e Aulas × Notas

```python
notas = pd.read_csv("/notas.csv")
notas
```

```python
X = notas[["livros", "aulas"]]
y = notas["notas"]
```

```python
modelo_notas = LinearRegression()
```

```python
modelo_notas.fit(X, y)
```

```python
modelo_notas.coef_
```

```python
modelo_notas.intercept_
```

```python
aluno1 = pd.DataFrame({"livros": [2], "aulas": [11]})

modelo_notas.predict(aluno1)
```

```python
aluno2 = pd.DataFrame({"livros": [0], "aulas": [5]})

modelo_notas.predict(aluno2)
```

```python
aluno3 = pd.DataFrame({"livros": [4], "aulas": [20]})

modelo_notas.predict(aluno3)
```

```python
aluno4 = pd.DataFrame({"livros": [2], "aulas": [10]})

modelo_notas.predict(aluno4)
```

```python
aluno5 = pd.DataFrame({"livros": [4], "aulas": [15]})

modelo_notas.predict(aluno5)
```

```python
y_pred_notas = modelo_notas.predict(X)
```

```python
mae_notas = mean_absolute_error(y, y_pred_notas)
mae_notas
```

```python
mse_notas = mean_squared_error(y, y_pred_notas)
mse_notas
```

```python
rmse_notas = mse_notas ** 0.5

rmse_notas
```

```python
r2_notas = r2_score(y, y_pred_notas)
r2_notas
```

```python
print("MAE:", mae_notas)
print("MSE:", mse_notas)
print("RMSE:", rmse_notas)
print("R²:", r2_notas)
```

```python
print("2 livros e 11 aulas:", modelo_notas.predict(pd.DataFrame({"livros": [2], "aulas": [11]}))[0])
print("0 livros e 5 aulas:", modelo_notas.predict(pd.DataFrame({"livros": [0], "aulas": [5]}))[0])
print("4 livros e 20 aulas:", modelo_notas.predict(pd.DataFrame({"livros": [4], "aulas": [20]}))[0])
print("2 livros e 10 aulas:", modelo_notas.predict(pd.DataFrame({"livros": [2], "aulas": [10]}))[0])
print("4 livros e 15 aulas:", modelo_notas.predict(pd.DataFrame({"livros": [4], "aulas": [15]}))[0])
```

# Classificação — Iris Flower com KNN

```python
from sklearn.datasets import load_iris
```

```python
iris_X, iris_y = load_iris(return_X_y=True)
```

```python
iris_X.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    iris_X,
    iris_y,
    test_size=0.2,
    random_state=42
)
```

```python
print("Treinamento:", X_train.shape)
print("Teste:", X_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred_iris = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
acuracia = accuracy_score(y_test, y_pred_iris)
acuracia
```

```python
print("Acurácia:", acuracia)
print("Acurácia (%):", acuracia * 100)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred_iris))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
matriz = confusion_matrix(y_test, y_pred_iris)
matriz
```

```python
iris = load_iris()
iris.target_names
```

```python
print(classification_report(
    y_test,
    y_pred_iris,
    target_names=iris.target_names
))
```
